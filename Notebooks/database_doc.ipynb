**PostgreSQL Dockerized Setup Documentation**

## **1. Overview**
This document serves as a reference for running a **PostgreSQL database in Docker**. The setup ensures a portable, persistent, and easily manageable database environment for development.

---

## **2. Running PostgreSQL in Docker**
To start a **PostgreSQL** container, use the following command:

```sh
docker run --name postgres-dev \
  -e POSTGRES_USER=admin \
  -e POSTGRES_PASSWORD=secret \
  -e POSTGRES_DB=mydb \
  -p 5432:5432 \
  -v postgres_data:/var/lib/postgresql/data \
  -d postgres
```

### **Command Breakdown:**
| **Flag**               | **Description** |
|------------------------|----------------|
| `--name postgres-dev`  | Names the container **postgres-dev** for easy reference. |
| `-e POSTGRES_USER=admin` | Sets the PostgreSQL **username** as `admin`. |
| `-e POSTGRES_PASSWORD=secret` | Sets the **password** to `secret`. |
| `-e POSTGRES_DB=mydb` | Creates a default database named `mydb`. |
| `-p 5432:5432` | Maps the container’s **PostgreSQL port (5432)** to the host machine. |
| `-v postgres_data:/var/lib/postgresql/data` | Uses a **Docker volume** (`postgres_data`) to persist database files. |
| `-d postgres` | Runs the container in **detached mode** (in the background). |

---

## **3. Accessing the Database**

### **Using psql (PostgreSQL CLI)**
If you have `psql` installed, connect to the database with:
```sh
psql -h 127.0.0.1 -U admin -d mydb
```
Enter the password `secret` when prompted.

### **Using pgAdmin**
1. Open pgAdmin.
2. Create a new server.
3. **Connection settings:**
   - **Host**: `127.0.0.1`
   - **Port**: `5432`
   - **Username**: `admin`
   - **Password**: `secret`

---

## **4. Managing the Container**

### **Check if PostgreSQL is Running**
```sh
docker ps
```

### **Stop the Container**
```sh
docker stop postgres-dev
```

### **Restart the Container**
```sh
docker start postgres-dev
```

### **Remove the Container** (⚠️ This deletes the database)
```sh
docker rm postgres-dev
```

### **Remove the Persistent Volume** (⚠️ This permanently deletes all data)
```sh
docker volume rm postgres_data
```

---

## **5. Backing Up & Restoring the Database**

### **Export Database (Backup)**
```sh
docker exec -t postgres-dev pg_dump -U admin -d mydb > backup.sql
```

### **Restore Database**
```sh
docker exec -i postgres-dev psql -U admin -d mydb < backup.sql
```

---

## **6. Notes & Considerations**
- **Data Persistence:** The database is stored in a Docker volume (`postgres_data`), meaning it **won't be lost** when restarting or stopping the container.
- **Environment Variables:** Change `POSTGRES_USER`, `POSTGRES_PASSWORD`, or `POSTGRES_DB` as needed, but ensure the application uses the correct credentials.
- **Connecting from an Application:** Use the database connection string:
  ```
  postgresql://admin:secret@127.0.0.1:5432/mydb
  ```
- **Cloud Migration:** If moving to a cloud provider (e.g., AWS RDS, GCP Cloud SQL), you can export data using `pg_dump` and restore it in the new instance.
